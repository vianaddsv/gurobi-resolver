```python
import gurobipy as gp
from gurobipy import GRB
```

```python
try:
    # 1. Cria o modelo (Se a licença tiver algum erro, vai falhar aqui)
    m = gp.Model("teste_licenca")
    
    # Reduz a verbosidade do log do Gurobi na tela (opcional)
    m.setParam('OutputFlag', 1)

    # 2. Cria as variáveis
    # x e y contínuas (por padrão o Gurobi já assume que são >= 0)
    x = m.addVar(vtype=GRB.CONTINUOUS, name="x")
    y = m.addVar(vtype=GRB.CONTINUOUS, name="y")

    # 3. Define a função objetivo: Maximizar x + y
    m.setObjective(x + y, GRB.MAXIMIZE)

    # 4. Adiciona a restrição: x + 2y <= 4
    m.addConstr(x + 2 * y <= 4, "restricao_1")

    # 5. Otimiza o modelo
    m.optimize()

    # 6. Verifica se encontrou a solução ótima e imprime
    if m.status == GRB.OPTIMAL:
        print("\n" + "="*40)
        print("🎉 SUCESSO! O Gurobi está funcionando!")
        print("="*40)
        print(f"Valor ótimo da função objetivo: {m.ObjVal}")
        print(f"Valor de x: {x.X}")
        print(f"Valor de y: {y.X}")
    else:
        print(f"\nO modelo não encontrou solução ótima. Código de status: {m.status}")

except gp.GurobiError as e:
    print(f"\n[ERRO DE LICENÇA OU GUROBI]: {e}")
except Exception as e:
    print(f"\n[ERRO DESCONHECIDO]: {e}")
```
